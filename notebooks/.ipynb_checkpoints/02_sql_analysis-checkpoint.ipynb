```python
import pandas as pd
from sqlalchemy import create_engine

df = pd.read_csv("outputs/final_dataset.csv")

#isimlendirmeler farklı olduğu için yeniden düzenledim
df['order_delivered_customer_date'] = pd.to_datetime(df['order_delivered_customer_date'])
df['order_estimated_delivery_date'] = pd.to_datetime(df['order_estimated_delivery_date'])
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])


df['delivery_time'] = (df['order_delivered_customer_date'] - df['order_purchase_timestamp']).dt.days
df['delay'] = (df['order_delivered_customer_date'] - df['order_estimated_delivery_date']).dt.days


engine = create_engine('postgresql://postgres:Pardon30.@localhost:5433/postgres')

# tablolar da düzenlendi tekrar
df_orders = df[['order_id', 'customer_unique_id', 'customer_city', 'customer_state', 
                'order_status', 'order_purchase_timestamp', 'year']]

# Tablo 2: 'delivery_time' ve 'delay' olarak güncelledim
df_performance = df[['order_id', 'total_price', 'total_freight', 'item_count', 
                     'top_category', 'review_score', 'delivery_time', 'delay']]

#SQL'e Gönder
df_orders.to_sql('dim_customers_orders', engine, schema='olist_project', if_exists='replace', index=False)
df_performance.to_sql('fact_order_performance', engine, schema='olist_project', if_exists='replace', index=False)

print("İşlem Başarılı! Kolon isimleri senkronize edildi ve PostgreSQL'e yüklendi.")
```

```python
from sqlalchemy import create_engine

# 1. Veritabanı Bağlantısı (Port: 5433)
engine = create_engine('postgresql://postgres:Pardon30.@localhost:5433/postgres')

# 2. Tablo 1: Müşteri ve Sipariş Detayları
df_orders = df[['order_id', 'customer_unique_id', 'customer_city', 'customer_state', 
                'order_status', 'order_purchase_timestamp', 'year']]

# 3. Tablo 2: Performans ve Finans (Senin kolon isimlerinle: delivery_time ve delay)
df_performance = df[['order_id', 'total_price', 'total_freight', 'item_count', 
                     'top_category', 'review_score', 'delivery_time', 'delay']]

# 4. PostgreSQL'e Gönder (olist_project şemasına)
df_orders.to_sql('dim_customers_orders', engine, 
                schema='olist_project', 
                if_exists='replace', 
                index=False)

df_performance.to_sql('fact_order_performance', engine, 
                      schema='olist_project', 
                      if_exists='replace', 
                      index=False)

print("Veriler PostgreSQL'e yüklendi!")
```
